## Series — Creation & Indexing

```python
s = pd.Series()
print(s)
print(type(s),len(s),id(s))
```

```python
arr_data= np.array(['apple','pineapple','lime','lemon'])
s=pd.Series(arr_data)
print(s)
print(type(s),len(s),id(s))
print(s[0],s[3])
```

```python
arr_data=np.array([100,200,300,500])
print(arr_data)
s=pd.Series(data=arr_data,copy=False)
print(s)
arr_data[0]=111
s[3]=555
print(arr_data)
print(s)
```

```python
arr_data=np.array([100,200,300,500])
print(arr_data)
s=pd.Series(data=arr_data,copy=True) #local allocation
print(s)
arr_data[0]=111
s[3]=555
print(arr_data)
print(s)
```

```python
arr_data= np.array(['apple','pineapple','lime','lemon'])
arr_index=[100,200,300,400]
s=pd.Series(data=arr_data,index=arr_index)
print(s)
print(s[100],s[400])
```

## DataFrame — Creation, Concatenation & Merge

```python
user_data=[['alice',19,'F','student'],['john',26,'M','student']]
user_column=['name','age','gender','job']
user1=pd.DataFrame(data=user_data,columns=user_column)
display(user1)
```

```python
user_data=dict(name=['alice','john','bob'],age=[19,26,22],gender=['F','M','M'],job=['student','manager','student'])
print(user_data)
user2=pd.DataFrame(data=user_data)
display(user2)
```

```python
user_data={'name':['eric','paul','peter'],'age':[19,26,22],'gender':['F','M','M'],'job':['scientist','engineer','doctor']}
print(user_data)
user3=pd.DataFrame(data=user_data)
display(user3)
```

```python
users=pd.concat([user1,user2,user3])
display(users)
```

```python
users=pd.concat([user1,user2,user3],ignore_index=True)
display(users)
```

```python
arr_data=users.to_numpy()
print(arr_data)
arr_data=np.array(users)
print(arr_data)
```

```python
dict_data=dict(name=['alice','john','eric','julie','smith'],height=[165,180,172,180,175])
print(dict_data)
user4=pd.DataFrame(data=dict_data)
display(user4)
```

```python
merge_inner=pd.merge(users,user4,on='name',how='inner')
display(merge_inner)
```

```python
merge_left=pd.merge(users,user4,on='name',how='left')
display(merge_left)
```

```python
merge_right=pd.merge(users,user4,on='name',how='right')
display(merge_right)
```

```python
merge_outer=pd.merge(users,user4,on='name',how='outer')
display(merge_outer)
```
